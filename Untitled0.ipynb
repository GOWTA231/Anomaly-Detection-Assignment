```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, LSTM, RepeatVector, TimeDistributed

# --- 1. Generate Synthetic Data (for demonstration) ---
# Create a sine wave with some noise
time_steps = np.linspace(0, 100, 1000)
data = np.sin(time_steps) + np.random.normal(0, 0.1, 1000)

# Inject intentional anomalies (spikes)
data[200:220] = 3.0  # Anomaly 1
data[600:605] = -3.0 # Anomaly 2

# Convert to DataFrame
df = pd.DataFrame(data, columns=['value'])

# --- 2. Preprocessing ---
scaler = MinMaxScaler()
df_scaled = scaler.fit_transform(df)

# --- 3. Method 1: Isolation Forest (Statistical) ---
print("Training Isolation Forest...")
# contamination=0.05 means we guess 5% of data is anomalous
iso_forest = IsolationForest(contamination=0.05, random_state=42)
df['anomaly_iso'] = iso_forest.fit_predict(df_scaled)
# Isolation Forest labels: -1 is anomaly, 1 is normal. Let's map -1 to True (Anomaly)
df['is_anomaly_iso'] = df['anomaly_iso'] == -1

# --- 4. Method 2: LSTM Autoencoder (Deep Learning) ---
print("Training LSTM Autoencoder...")

# Function to create sequences for LSTM [samples, time_steps, features]
def create_sequences(values, time_steps=10):
    output = []
    for i in range(len(values) - time_steps):
        output.append(values[i : (i + time_steps)])
    return np.stack(output)

TIME_STEPS = 30
X_lstm = create_sequences(df_scaled, TIME_STEPS)

# Build the model
model = Sequential([
    LSTM(64, activation='relu', input_shape=(X_lstm.shape[1], X_lstm.shape[2]), return_sequences=False),
    RepeatVector(X_lstm.shape[1]),
    LSTM(64, activation='relu', return_sequences=True),
    TimeDistributed(Dense(X_lstm.shape[2]))
])

model.compile(optimizer='adam', loss='mse')
model.fit(X_lstm, X_lstm, epochs=10, batch_size=32, validation_split=0.1, verbose=0)

# Predict reconstruction
X_pred = model.predict(X_lstm)
# Calculate Mean Absolute Error (reconstruction error)
test_mae_loss = np.mean(np.abs(X_pred - X_lstm), axis=1)

# Set a threshold for anomaly (e.g., anything above 0.2 error)
THRESHOLD = 0.25
# Pad the beginning because sequence generation shortens the array
padding = np.zeros((TIME_STEPS, 1))
# Combine padding with results to match original length
anomaly_scores = np.concatenate([padding, test_mae_loss])
df['is_anomaly_lstm'] = anomaly_scores > THRESHOLD

# --- 5. Visualization ---
plt.figure(figsize=(15, 6))
plt.plot(df.index, df['value'], label='Data', color='grey', alpha=0.6)

# Highlight Isolation Forest Anomalies (Red X)
anomalies_iso = df[df['is_anomaly_iso']]
plt.scatter(anomalies_iso.index, anomalies_iso['value'], color='red', label='Isolation Forest', marker='x', s=50)

# Highlight LSTM Anomalies (Blue Circle)
anomalies_lstm = df[df['is_anomaly_lstm']]
plt.scatter(anomalies_lstm.index, anomalies_lstm['value'], color='blue', label='LSTM Autoencoder', marker='o', facecolors='none', s=100)

plt.title("Anomaly Detection: Isolation Forest vs LSTM Autoencoder")
plt.legend()
plt.show()

print("Mission Complete. Anomalies plotted.")
```
